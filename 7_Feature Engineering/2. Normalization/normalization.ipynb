# Normalization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('wine_data.csv',usecols=[0,1,2])
df.columns = ['class_lebel','alcohol','Malic Acid']
df.head()
```

```python
from sklearn.model_selection import train_test_split as tts

x_train,x_test,y_train,y_test = tts(df.drop('class_lebel',axis = 1),
                                   df['class_lebel'],
                                   test_size = 0.3,
                                   random_state = 0)
x_train.shape,x_test.shape
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
scaler.fit(x_train)
x_train_scaled = scaler.transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled = pd.DataFrame(x_train_scaled,columns = x_train.columns)
x_test_scaled = pd.DataFrame(x_test_scaled,columns = x_test.columns)
x_train_scaled.describe()
```

```python
x_test_scaled.head()
```

```python
fig,(ax1,ax2) = plt.subplots(ncols = 2, figsize=(12, 5))
ax1.scatter(x_train['alcohol'],x_train['Malic Acid'])
ax1.set_title('before nomalisation')
ax2.scatter(x_train_scaled['alcohol'],x_train_scaled['Malic Acid'])
ax2.set_title('after nomalisation')
```

```python
fig , (ax1,ax2) = plt.subplots(ncols = 2,figsize = (12,5))
ax1.set_title("before Scaling")
sns.kdeplot(x_train['alcohol'],ax=ax1)
sns.kdeplot(x_train['Malic Acid'],ax = ax1)
ax2.set_title("after scaling")
sns.kdeplot(x_train_scaled['alcohol'],ax=ax2)
sns.kdeplot(x_train_scaled['Malic Acid'],ax = ax2)
```

```python
fig , (ax1,ax2) = plt.subplots(ncols = 2, figsize = (12,5))
ax1.set_title("Before Normalization")
sns.kdeplot(x_train['alcohol'],ax=ax1)
ax2.set_title("After Normalization")
sns.kdeplot(x_train_scaled['alcohol'],ax=ax2)
```

```python
fig ,(ax1,ax2) = plt.subplots(ncols=2,figsize = (12,5))
ax1.set_title("Before Normalization")
sns.kdeplot(x_train['Malic Acid'],ax=ax1)
ax2.set_title("After Normalization")
sns.kdeplot(x_train_scaled['Malic Acid'],ax=ax2)
```
