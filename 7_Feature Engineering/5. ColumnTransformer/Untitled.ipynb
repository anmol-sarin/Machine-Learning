```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('covid_toy.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(df.drop('has_covid',axis = 1),
                                                       df['has_covid'],
                                                       test_size = 0.2 ,
                                                       random_state = 0)
x_train.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder(categories = [['Mild','Strong']])
new_x_train1 = oe.fit_transform(x_train[['cough']])
new_x_test1 = oe.transform(x_test[['cough']])
new_x_train.shape
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(drop = 'first' , dtype = np.int32)
new_x_train2 = ohe.fit_transform(x_train[['gender','city']])
new_x_test2 = ohe.transform(x_test[['gender','city']])
new_x_train2.shape
```

```python
from sklearn.impute import SimpleImputer
si = SimpleImputer()
new_x_train3 = si.fit_transform(x_train[['fever']])
new_x_test3 = si.transform(x_test[['fever']])
new_x_train3.shape
```

```python
from sklearn.compose import ColumnTransformer
# syntax of ColumnTransformer
# {Variable} = ColumnTransformer(transformers = [
#({nameof diff transformers},{name of function used},{name of fields}),
#({nameof diff transformers},{name of function used},{name of fields}),
#({nameof diff transformers},{name of function used},{name of fields}),]
#, remainder = {name of fields not being processed })
```

```python
transformer = ColumnTransformer(transformers =[
    ('tnf1',SimpleImputer(),['fever']),
    ('tnf2',OrdinalEncoder(categories = [['Mild' , 'Strong']]),['cough']),
    ('tnf3',OneHotEncoder(drop = ('first')),['gender' , 'city'])
],remainder='passthrough')
```
